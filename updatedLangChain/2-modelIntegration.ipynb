### Streaming and Batch

#### Streaming

Most models can stream their output content while it is being generated . By displaying output progressively , streaming significantly improves the user experience , particularly for longer responses . Calling stream() particullarly returns an iterators the often yeilds output chunks as they are produced. You can use a loop to process each chunk in real time.

```python
import os 
from langchain_google_genai import ChatGoogleGenerativeAI

model=ChatGoogleGenerativeAI(model = "gemini-3.6-flash")
model.stream("Write a 100 word essay about AI")
for chunk in model.stream("Write a 100 word essay about AI"):
    print(chunk.text,end="\n",flush=True)
```

### Batch


Batching is the collection of independent request to a model that can significantly improve performance and reduce cost , as the processing can be done in parallel.

```python
import os 
from langchain_google_genai import ChatGoogleGenerativeAI

model=ChatGoogleGenerativeAI(model = "gemini-3.6-flash")

responses = model.batch([
    "Why do parrot have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing"
])

for response in responses:
    print(response)
```
